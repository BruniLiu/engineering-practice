id怎么来的 idx2word和word2idx是哪来的

```python
from collections import Counter

toy = ['the', 'cat', 'sat', 'the', 'dog', 'sat', 'the', 'cat', 'ran', 'the', 'cat', 'zzz']
c = Counter(toy)
print("1. 词频：", c)

kept = [(w, n) for w, n in c.most_common() if n >= 2]
print("2. 保留：", kept)

idx2word = [w for w, _ in kept]
print("3. idx2word:", idx2word)

print(idx2word[0])   # the
print(idx2word[1])   # cat
print(idx2word[2])   # sat

word2idx = {w: i for i, w in enumerate(idx2word)}
print("4. word2idx:", word2idx)
```
