## 关于哈希表

哈希表主要想解决的问题是这个东西是否存在

如果有10000个单词，要反复查king这个单词是不是在这里面，如果用数组存要一遍一遍的扫，O(n)

### 核心想法：让数据自己告诉你它该在哪儿

如果我要存的是 0 到 999 的整数，事情就简单了——开一个长度 1000 的数组，数字 n 就存在下标 n 的位置。查 n 在不在？看一眼 arr[n]，一次访问，O(1)。

这个思路完美，但有个致命限制：下标必须是有限范围内的整数。而我们要存的是字符串，字符串没有下标。

## 哈希函数

比如一个最简单的哈希函数：把每个字符的 ASCII 码加起来。

"cat" → 99 + 97 + 116 = 312

然后对数组长度取模，得到下标。数组长度假设是 100：

312 % 100 = 12

但是这么干会存在哈希冲突，所以需要解决方案

```python
print(hash('cat'))
print(hash('act'))
print(hash('cau'))
```
