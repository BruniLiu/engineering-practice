## About Path  关于路径

这一步仅仅是创建变量，不创建任何全新的文件夹

```python
from pathlib import Path

DATA_DIR = Path("data")
print(DATA_DIR)
```

真正创建文件夹

```python
DATA_DIR.mkdir(exist_ok=True)
```

现在要把文件放进你要放的位置

```python
TEXT8_ZIP = DATA_DIR / "text8.zip"
TEXT8_TXT = DATA_DIR / "text8"

print(DATA_DIR, DATA_DIR.exists())
print(TEXT8_ZIP, TEXT8_ZIP.exists())
print(TEXT8_TXT, TEXT8_TXT.exists())
```
